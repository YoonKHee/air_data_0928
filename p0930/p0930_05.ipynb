```python
import pandas as pd

df =pd.read_excel('file/score.xlsx',index_col='지원번호')
```

```python
# df.columns 컬럼리스트
df.columns
# df['컬럼이름']
df['이름']
# df[['컬럼이름','컬럼이름']]: 2개 이상일때는 []리스트형태로
df[['이름','수학']]
# df['이름':'과학'] # 슬라이싱 안됨
df[df.columns[0:4]] # columns를 사용해서 슬라이싱 적용
df.loc[:,'이름':'국어'] # loc 를 사용해서 슬라이싱
df[0:3] # index에 해당되는 3개 출력
df.iloc[0:3]  # index에 해당되는 3개 출력

df[df.columns[-1]] # 마지막 컬럼 출력
df[df.columns[[0,3,-1]]]
```

```python
# loc 출력 - index 출력
# loc [[index이름:],[컬럼이름:]]
df.loc['1번']
df.loc[['1번','2번']]
df.loc['1번':'5번']
df.loc['1번','이름':'수학'] # 슬라이싱
df.loc['1번':'5번','이름':'수학'] # 슬라이싱
df.loc[['1번','5번'],'이름':'수학'] # 슬라이싱
```

```python
# iloc 출력 - 리스트 번호로 출력
df.iloc[0]
df.iloc[0:4]
df.iloc[[0,1,4]]
df.iloc[0:4,1:5]
df.iloc[[0,4],1:5]
```

```python
# 조건 : True, False 
# df['키']>=185
df[df['키']>=185]['키']
df[df['키']>=80]['학교']
```

```python
#df.loc[(index)row]
df.loc[df['키']>190, '학교':'과학']
```

```python
df.loc[(df['키']>=185)&(df['학교']=='디지털고')]
df.loc[(df['키']>=185)|(df['학교']=='디지털고')]
df.loc[(df['국어']>85)|(df['수학']>=90)]
```
