### 문자열 함수.str

```python
# 문자열함수 str함수 - startswith:'강'으로 시작하는 문자열
# startswith, endswith,contains,lower,upper, capitalize


# df[df['이름'].str.startswith('강')]
# df.loc[df['이름'].str.startswith('강')]

# contains : 포함되어 있는. ~은 not
# df[df['이름'].str.contains('원')]
# df[~df['이름'].str.contains('원')] # ~을 not

# lower,upper,capitalize
df['SW특기'].str.lower() # 소문자
df['SW특기'].str.upper() # 대문자
df['SW특기'].str.capitalize() # 첫 글자만 대문자
df[df['SW특기'].str.startswith('J')] #J로 시작하는 것

#isin : 여러개 검색 ['python','java']
# langs = ['python','java']
# #SW특기 값 소문자로 변경후 리스트 검색
# df[df['SW특기'].str.lower().isin(langs)] # 안의 것을 소문자로 다 바꿔주고 langs의 것들을 찾아달라는 말

#
df['SW특기'].str.contains('Java',na=False) # na = False :NaN 데이터 출력하지 않음
df['SW특기'].str.contains('Java',na=True) # na = True :NaN 데이터 출력
df[df['SW특기'].str.lower().str.contains('java',na=True)] # 소문자로 변경후 검색
# df['SW특기']

# unique 중복제거
df['학교'].unique()
```
