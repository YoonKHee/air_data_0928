```python
# 그래프를 파악하는 용도, 그리는 용도가 아님
x = ['Jan',"Feb",'Mar']
y = [5,15,30] 
plt.title('꺾은선 graph') # 상단 타이틀 / 한글 사용하려면 폰트 패밀리 적용, 사이즈는 폰트 사이즈.
plt.bar(x,y) #그래프 그리기
plt.show() # 설명글자 부분 사라짐
```
