```python
df = pd.read_csv('file/namu2.csv')
df = df.head(3)
df
```

```python
def text_cleaning(text):
    if pd.isna(text):
        return ''
    hangul = re.compile('[^ ㄱ-ㅣ가-힣]+') # 정규 표현식에 해당이 되는지 확인
    return hangul.sub('',text) # sub는 빼다. 라는 뜻
```

```python
df['content'] = df['content'].apply(lambda x:text_cleaning(x))
df['content']
contents = " ".join(df['content'])

contents
```

```python
from konlpy.tag import Okt
n_tag = Okt()
nouns = n_tag.nouns(df['content'][0])
count = Counter(nouns)
count
```

```python
# 1개단어 키워드 제외
# count # 딕셔너리 타입
# for x in count:
#     if len(x)>1:
#         {x:count[x]}
remove_counter = Counter({x:count[x] for x in count if len(x)>1}) # 위와 같은뜻
remove_counter
```
