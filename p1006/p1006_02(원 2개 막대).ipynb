```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic' # 한글처리 window 맑은고딕이 디폴트로 저장되어있음
# matplotlib.rcParams['font.family'] = 'AppleGothic' # 한글처리 mac 애플고딕이 디폴트로 저장되어있음
matplotlib.rcParams['font.size'] = 15
matplotlib.rcParams['axes.unicode_minus'] = False # -(마이너스)부호 처리
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('file/students.xlsx')
school=df['학교'].value_counts(normalize=True)
school.values
school.index
explode = [0.1,0,0]
color = ['red','green','blue']
plt.pie(school.values, labels=school.index, explode=explode,autopct='%.1f%%',startangle=90,counterclock=False,colors=color)
plt.show()
```

```python
values=[30,25,20,13,10,2]
labels = ['파이썬','html','자바스크립트','판다스','그래프','디장고']
# plt.bar(labels,values)
explode = [0.1,0.2,0,0,0,0] # 원파트 띄우기
# startangle 기준각도, counterclock :시계방향
# autopct = '%.1f%%'
plt.pie(values,labels=labels,startangle=90,counterclock=False,explode=explode,autopct='%.1f%%')
plt.title('언어별 선호도')
plt.legend(loc=(1.2,0.3),title='언어별 선호도')
plt.show()
```

```python
# # beer,spirit,wine => 합계 = total_serving
# # 합계의 대한 막대그래프를 출력 높은순으로 정렬
# # 한국 맏ㄷ 그래프를 red에 추가해서 출력
df['total_servings']=df['beer_servings']+df['spirit_servings']+df['wine_servings']

sort_df = df.sort_values('total_servings',ascending=False) 
korea = sort_df[sort_df['country'] == 'South Korea']


df['total_servings']=df['total_servings'].sort_values(ascending=False)
x = sort_df['country']
y = sort_df['total_servings']
x2 = korea['country']
y2 = korea['total_servings']
plt.bar(x,y,label='전체')
plt.bar(x2,y2,label='한국')
plt.legend()
# plt.xticks(x,idx)
plt.show()
```
