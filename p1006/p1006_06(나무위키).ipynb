```python
# 3. 파일 BeautifulSoup변환
with open('file/namu.html','r',encoding='utf-8') as f:
    soup = BeautifulSoup(f,'lxml')


divs = soup.find_all('div',{'class':'M5XHy6jV BqyggfHC'})
# divs[0]

# dataFrame으로 변환해서 저장하시오 # namu_link.csv

# "https://namu.wiki/w/%EB%8D%B0%EC%A7%80%EB%A0%88%20%EB%91%90%EC%97%90"

a = []
for i in range(len(divs)):
    name= divs[i].find('a').get_text(strip=True)
    url= divs[i].find('a')['href']
    url = 'https://namu.wiki'+url
    # print(f'{i}.{name},{url}')
    a.append({
        'title':name,
        'url':url
    })
# # print(len(divs))
# print(a)
a

df = pd.DataFrame(a)
df

df.to_csv('file/namu02_link.csv',encoding='utf-8-sig')
```
