```python
# 키 +cm 추가
# df['키'] =df['키'] +'cm' # 에러남 int와 str 문자열이 서로 다름 타입이 다름
# df['키'] = df['키'] +100 # 같은 int는 됨 같은 문자열 타입은 된다
# 타입이 다른경우 타입 변경후 적용
df['키']=df['키'].astype(str) +'cm' # str타입으로 변환 후 적용
```

```python
# 함수 적용 : apply() - 컬럼 전체에 적용
def add_cm(h):
    return str(h)+'센티미터'

df['키'] = df['키'].apply(add_cm)
```

```python
# 함수적용
def capChange(lang):
    if pd.notnull(lang):
        return lang.lower()
    else: return '특기없음'

df['SW특기']=df['SW특기'].apply(capChange)


# df['SW특기'].str.lower()
```

```python
# 1. 합계 : 과목 총점을 계산에서 컬럼 추가
# 2. 평균 컬럼을 추가
# 3. 등급 컬럼을 추가해서 90점 이상A, 80이상 B,70이상C,60이상D 그외 f를 입력
# 3번은 함수를 사용할것
def grade(avg):
    if avg>=90 :
        return 'A'
    elif avg>=80 :
        return 'B'
    elif avg>=70 :
        return 'C'
    elif avg>=60 :
        return 'D'
    else: return 'F'

df['합계'] = df['국어']+df['영어']+df['수학']+df['과학']+df['사회']
df['평균'] = df['합계']/5
df['등급'] = df['평균'].apply(grade)
df
```
