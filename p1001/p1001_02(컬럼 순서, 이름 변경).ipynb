```python
import pandas as pd

df = pd.read_excel('file/score.xlsx',index_col='지원번호')
```

```python
# 합계,평균,결과 컬럼 추가
# 결과 평균점수가 70점 이상은 합격, 불합격으로 출력
df['합계'] =df['국어']+df['영어']+df['수학']
df['평균'] =df['합계']/3
df['결과'] = "불합격"
df.loc[df['평균']>=70, '결과'] = '합격'

```

```python
# cols리스트 순서를 변경
# 1. 리스트타입으로 변경
cols= list(df.columns) # 리스트 타입으로 변경후 순서를 변경


cols[-2:] # 평균,결과 컬럼
cols[:-2] # 이름~합계 컬럼
cols[0] # 이름
type(cols[:-2]) # list
type(cols[0]) # str


# 2. 리스트 순서를 변경
cols=[cols[0]]+cols[-2:]+cols[:-2] # str+list+list 0>error, 그래서 str을 []list로 변환해야함

# 3.변경된 리스트 적용
df=df[cols]
df
```

```python
# 컬럼수정 : replace(), 존재하는 컬럼에 값을 입력하면 수정됨
# 컬럼추가 : 없는 컬럼에 값을 입력하면 추가됨
# 컬럼삭제 : drop()형태를 사용 여러개 삭제 가능,inplace=True 적용해야함( del은 1개만 삭제됨.)
```

```python
# row : index 수정 추가 삭제
# index 삭제 - 조건 사용해서 삭제
# 수학 점수가 80점 미만 학생 삭제

filt = df['수학']>=80
# df[~filt].count()
# df2 = df[~filt]
# df2

# # column삭제
# df.drop(columns=['평균','결과'],inplace=True) 

# index row삭제
# df.drop(index=['1번','2번'],inplace=True) 

# 2. 조건을 이용한 index삭제
# filt.index
df.drop(index=df[~filt].index,inplace=True) 
df
```
