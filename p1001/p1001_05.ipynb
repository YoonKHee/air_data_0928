```python
import pandas as pd
df = pd.read_excel('file/score.xlsx',index_col='지원번호')
```

```python
import random
df['요일'] = 0
# list = [0,1,2,3,4,5,6]
for i in range(8):
    df.iloc[i,9] = random.randint(0,6)


df
```

```python
# 0-일 1-월 2-화 3-수 4-목 5-금 6-토
# 출력하시오
def today(a):
    if a == 0:
        return "일요일"
    elif a == 1:
        return "월요일"
    elif a == 2:
        return "화요일"
    elif a == 3:
        return "수요일"
    elif a == 4:
        return "목요일"
    elif a == 5:
        return "금요일"
    elif a == 6:
        return "토요일"
df['요일']=df['요일'].apply(today)
df
```

```python
data = {
    '번호':['1번','2번','3번','4번','5번','6번','7번','8번'],
    '이름':['홍길동','유관순','이순신','강감찬','김구','김유신','홍길자','홍길순'],
    '달러':['$19,700','$18,400','$16,800','$18,700','$18,800','$20,200','$18,800','$19,000'],
    '우대':[90,40,80,40,15,80,55,100]
}
df = pd.DataFrame(data)
df.set_index('번호',inplace=True)
df
```

```python
# 1 금액컬럼을 추가하시오 (금액:달러+(달러*달러우대*0.01))
# 2 우대가 50 이상인 사람에게만 달러+달러*우대*0.01
#50 미만은 달러만 입력

# 우대 50 이상인 사람에게 달러+달러X우대X0.01
# 우대가 50미만 달러

df['달러']=df['달러'].str.replace({'$':'', ',':''}).str.strip().astype(int)

def cal_dal(a):
    if a >=50:
        return a
    else:
        return 0
df['우대']=df['우대'].apply(cal_dal)
df['금액'] = df['달러']+(df['달러']*df['우대']*0.01)



def cal_change(df2):
    if df2['우대']>=50:
        return df2['달러']+(df2['달러']*df2['우대']*0.01)
    else:
        return df2['달러']





df
```
