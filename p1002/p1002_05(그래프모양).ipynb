```python
# score.xlsx 국어 점수를 선그래프로 출력하시오.
# 상단에는 값을 출력

df = pd.read_excel('file/score.xlsx',index_col='지원번호')
df

x = df.index
y = df['국어'] # 타입이 pandas타입이라 y[0]이렇게 꺼낼수가 없음
y = list(df['국어']) # 그래서 리스트로 묶어서 내보내야함
plt.plot(x,y,linewidth=2,marker='o',ms=10,mfc='yellow',linestyle='-.',\
        color='blue')
for i,v in enumerate(y):
    plt.text(x[i],y[i]+0.7,v,ha='center',color='black') # 판다스로 가져왔을때 리스트타입으로 바꿔야지 가능
    # plt.text(x[i],v+0.7,v,ha='center',color='black') # 판다스타입일 경우 v가 y의 value값 이므로 이므로 v를 써야함

plt.show()
```
