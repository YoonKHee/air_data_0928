```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic' # 한글처리 window 맑은고딕이 디폴트로 저장되어있음
# matplotlib.rcParams['font.family'] = 'AppleGothic' # 한글처리 mac 애플고딕이 디폴트로 저장되어있음
matplotlib.rcParams['font.size'] = 15
matplotlib.rcParams['axes.unicode_minus'] = False # -(마이너스)부호 처리
```

```python
# 축설명글 색상, 위치
x = [1,2,3]
y = [10,15,30]

# 범례추가, 범례위치
# linewidth : 선 두께, marker :x,s,o,^, linestyle : 선모양(None,'--',':','-.')
# color:선색상
# markersize(ms):마커 크기조정, markerfacecolor(mfc):마커 색상, markeredgecolor(mec): 마커 테두리 색상, alpha : 투명도
# plt.plot(x,y,linewidth=2,marker='^',label = 'zz',linestyle = None, ms=10,mfc='green',mec='red',alpha=0.5)       # 그래프 그리기 / 설정부분보다 위쪽에 배치
plt.plot(x,y,linewidth=2,marker='o', ms=10,mfc='green',linestyle='-.',\
        color='pink')       # 그래프 그리기 / 설정부분보다 위쪽에 배치
# plt.scatter(x,y) 
# plt.legend(loc='upper right') # 위쪽 오른쪽편 위치이동 입력값이 없으면선에 걸리지 않게 자동배치됨
# upper conter lower / left conter right
plt.legend()
plt.title('꺾은선 graph')  # 상단타이틀

# X선 설명부분
plt.xlabel('매출액',color='red',loc='right') # X축설명글 색상, 위치
plt.ylabel('월별',color='#00aa00',loc='top') #Y축설명글 색상, 위치


# # 그래프 간격조정 ( 입력값 형태로 구성 )
# plt.xticks([1,1.2,1.4,1.6,2,3])
# plt.yticks([-15,10,20,30]) #간격부분 조정을 하면 꺾이는 상태를 조정
# plt.yticks([-15,10,20,30,100]) #간격부분 조정을 하면 꺾이는 상태를 조정
plt.show()
```
