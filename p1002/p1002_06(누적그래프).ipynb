```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic' # 한글처리 window 맑은고딕이 디폴트로 저장되어있음
# matplotlib.rcParams['font.family'] = 'AppleGothic' # 한글처리 mac 애플고딕이 디폴트로 저장되어있음
matplotlib.rcParams['font.size'] = 15
matplotlib.rcParams['axes.unicode_minus'] = False # -(마이너스)부호 처리
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('file/score.xlsx',index_col='지원번호')
df
```

```python
df.loc['1번','국어']
df.loc['1번','영어']
df.loc['1번','수학']
```

```python
# 선 스타일
x = df['이름']
x_1= list(x)
y1 = df['국어']
y_1 = list(y1)
y2 = df['영어']
y_2 = list(y2)
y3 = df['수학']
y_3 = list(y3)
plt.figure(figsize=(10,5))

plt.title('학생성적그래프')  # 상단타이틀
plt.bar(x,y1,label='국어',width=0.5)
plt.bar(x,y2,bottom=y1,label='영어',width=0.5)
plt.bar(x,y3,bottom=y1+y2,label='수학',width=0.5)
# ncol:범례가 1줄로 나타남
plt.legend(ncol=3)
plt.grid(axis='y',linestyle=':',alpha=0.5)

for i,(v,v2,v3) in enumerate(zip(y1,y2,y3)):
    plt.text(i,v/2,v,ha='center',va='center')
    plt.text(i,v+v2/2,v2,ha='center',va='center')
    plt.text(i,v+v2+v3/2,v3,ha='center',va='center')




# for i,v in enumerate(y1):
#     plt.text(x_1[i],y_1[i]+2,v,va='center')
# for i,v in enumerate(y2):
#     plt.text(x_1[i],y_2[i]+y_1[i]+2,v,va='center')
# for i,v in enumerate(y3):
#     plt.text(x_1[i],y_2[i]+y_1[i]+y_3[i]+2,v,va='center')


plt.show()
```
