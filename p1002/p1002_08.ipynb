```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic' # 한글처리 window 맑은고딕이 디폴트로 저장되어있음
# matplotlib.rcParams['font.family'] = 'AppleGothic' # 한글처리 mac 애플고딕이 디폴트로 저장되어있음
matplotlib.rcParams['font.size'] = 15
matplotlib.rcParams['axes.unicode_minus'] = False # -(마이너스)부호 처리
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('file/students.xlsx',index_col='번호')
df
```

```python
# students.xlsx 파일을 읽어와서
# 학교별 국어,영어,수학평균을 막대그래프로 출력하시오.
a=df.groupby(['학교'])[['국어','영어','수학']].mean()
a =a.round(2)
# index = [1,2,3]
index = np.arange(a.shape[0])
y=a['국어']
y2=a['영어']
y3=a['수학']
plt.title('학교별 국어,영어,수학 평균')
plt.figure(figsize=(10,5))
colors = ['red','blue','green']
plt.bar(index,y,label='국어',width=0.2)
plt.bar(index+0.2,y2,label='영어',width=0.2)
plt.bar(index+0.4,y3,label='수학',width=0.2)

for i,(v,v2,v3) in enumerate(zip(y,y2,y3)):
    plt.text(i,v+2,v, va='center',ha='center')
    plt.text(i+0.2,v2+2,v2, va='center',ha='center')
    plt.text(i+0.4,v3+2,v3, va='center',ha='center')

# 학교별, 학년별 국어,영어,수학 평균을 출력하시오.
df.groupby(['학교'])[['국어','영어','수학']].mean()
df.groupby(['학교','학년'])[['국어','영어','수학']].mean()



# 합계,평균,등수 컬럼을 추가하고 값을 입력하시오.
df['합계'] = df['국어']+df['영어']+df['수학']
df['평균'] = df['합계']/3
df['등수'] = df['평균'].rank(ascending=False,method='min').astype(int)
df.sort_values('등수',ascending=True, )
```
