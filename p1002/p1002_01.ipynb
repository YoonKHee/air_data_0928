```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic' # 한글처리 window 맑은고딕이 디폴트로 저장되어있음
# matplotlib.rcParams['font.family'] = 'AppleGothic' # 한글처리 mac 애플고딕이 디폴트로 저장되어있음
matplotlib.rcParams['font.size'] = 15
matplotlib.rcParams['axes.unicode_minus'] = False # -(마이너스)부호 처리
```

```python
# 그래프를 파악하는 용도, 그리는 용도가 아님
x = ['Jan',"Feb",'Mar']
y = [5,-15,30] 
plt.title('꺾은선 graph') # 상단 타이틀 / 한글 사용하려면 폰트 패밀리 적용, 사이즈는 폰트 사이즈.
plt.plot(x,y) #그래프 그리기
plt.show() # 설명글자 부분 사라짐
```

```python
# 축 설명글 생상, 위치
x = ['Jan',"Feb",'Mar']
y = [-15,5,60] 
# 범례
plt.plot(x,y,label= '범례(매출액)')
plt.legend()
plt.title('꺾은선 graph') # 상단 타이틀 / 한글 사용하려면 폰트 패밀리 적용, 사이즈는 폰트 사이즈.

# X선 설명부분
plt.xlabel('매출액',color= 'red', loc='right') # x축 설명글 색상, 위치
plt.ylabel('월별',color='green', loc='top') # y축 설명글 색상, 위치





#그래프 그리기
plt.plot(x,y) 

#그래프 간격조정 (입력값 형태로 구성)//설정부분볻 위쪽에 배치.
plt.xticks([1,1.2,1.4,1.6])
plt.yticks([-25,-5,15,30,40,60,100])#간격 부분을 조정을 하면 꺾이는 상태를 조정.

plt.show() # 완성된 그래프 보여주기
```

```python
# 축 설명글 생상, 위치
x = ['1번',"2번",'3번']
y = [80,60,90] 
y2 = [40,50,100] 
y3 = [80,85,100] 
# 범례

plt.plot(x,y,label='국어점수')
plt.plot(x,y2,label='영어점수')
plt.plot(x,y3,label='수학점수')
plt.legend()

plt.title('꺾은선 graph') # 상단 타이틀 / 한글 사용하려면 폰트 패밀리 적용, 사이즈는 폰트 사이즈.

# X선 설명부분
plt.xlabel('매출액',color= 'red', loc='right') # x축 설명글 색상, 위치
plt.ylabel('월별',color='green', loc='top') # y축 설명글 색상, 위치





#그래프 그리기
plt.plot(x,y) 

# #그래프 간격조정 (입력값 형태로 구성)//설정부분볻 위쪽에 배치.
# plt.xticks([0,1,20,30,40,50,60,70,80,90,100])
# plt.yticks([1,2,3,4])#간격 부분을 조정을 하면 꺾이는 상태를 조정.

plt.show() # 설명글자 부분 사라짐
```

```python
# 축설명글 색상, 위치
x = [1,2,3]
y = [-15,10,30]

# 범례추가, 범례위치
plt.plot(x,y,label='국어점수')       # 그래프 그리기 / 설정부분보다 위쪽에 배치
# plt.legend(loc='upper right') # 위쪽 오른쪽편 위치이동 입력값이 없으면선에 걸리지 않게 자동배치됨
# upper conter lower / left conter right
plt.legend(loc=(0.5,0.5))
plt.title('꺾은선 graph')  # 상단타이틀

# X선 설명부분
plt.xlabel('매출액',color='red',loc='right') # X축설명글 색상, 위치
plt.ylabel('월별',color='#00aa00',loc='top') #Y축설명글 색상, 위치


# 그래프 간격조정 ( 입력값 형태로 구성 )
plt.xticks([1,1.2,1.4,1.6,2,3])
plt.yticks([-15,10,20,30]) #간격부분 조정을 하면 꺾이는 상태를 조정
# plt.yticks([-15,10,20,30,100]) #간격부분 조정을 하면 꺾이는 상태를 조정
plt.show()
```
